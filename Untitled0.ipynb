```python
from ultralytics.data.utils import check_det_dataset

data_info = check_det_dataset("VOC.yaml")
voc_root = str(data_info["path"])

voc_split_config = {
    "path": voc_root,
    "train": [
        "images/train2012",
        "images/train2007",
        "images/val2012",
    ],
    "val": [
        "images/val2007",
    ],
    "test": [
        "images/test2007",
    ],
    "names": data_info["names"],
}

YAML_PATH = "/content/voc_split.yaml"
with open(YAML_PATH, "w", encoding="utf-8") as f:
    yaml.dump(voc_split_config, f, sort_keys=False)

def count_images(rel_paths):
    total = 0
    for p in rel_paths:
        full = os.path.join(voc_root, p)
        if os.path.exists(full):
            total += len([x for x in os.listdir(full) if x.endswith((".jpg", ".png"))])
    return total

print("Veri seti yolu:", voc_root)
print(f"Train görsel sayısı: {count_images(voc_split_config['train'])}")
print(f"Val görsel sayısı:   {count_images(voc_split_config['val'])}")
print(f"Test görsel sayısı:  {count_images(voc_split_config['test'])}")
print("Sınıflar :", list(data_info["names"].values()))
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

start_time = time.time()

results = model.train(
    data=YAML_PATH,
    epochs=20,
    imgsz=640,
    batch=32,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,
    patience=8,
    seed=42,
    workers=2,
    project=SAVE_DIR,
    name="yolov8n_voc_run",
    exist_ok=True,
    plots=True,
)

train_seconds = time.time() - start_time
print(f"\nToplam Eğitim Süresi: {train_seconds / 60:.2f} dakika ({train_seconds:.1f} saniye)")
```

```python
import os
import pandas as pd
from ultralytics import YOLO
from IPython.display import Image, display

SAVE_DIR = "/content/drive/MyDrive/cv_yolov8_voc"
RUN_DIR = os.path.join(SAVE_DIR, "yolov8n_voc_run")
BEST_WEIGHTS = os.path.join(RUN_DIR, "weights", "best.pt")
YAML_PATH = "/content/voc_split.yaml"

assert os.path.exists(BEST_WEIGHTS), f"Model bulunamadı: {BEST_WEIGHTS}"
assert os.path.exists(YAML_PATH), f"YAML bulunamadı: {YAML_PATH}"

best_model = YOLO(BEST_WEIGHTS)

test_metrics = best_model.val(
    data=YAML_PATH,
    split="test",
    imgsz=640,
    batch=32,
    project=SAVE_DIR,
    name="yolov8n_voc_test",
    exist_ok=True,
)

speed = test_metrics.speed
total_ms = speed["preprocess"] + speed["inference"] + speed["postprocess"]

csv_path = os.path.join(RUN_DIR, "results.csv")
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    df.columns = df.columns.str.strip()
    print(f"Tamamlanan Epoch Sayısı: {len(df)}")
    if "time" in df.columns:
        print(f"Kayıtlı Eğitim Süresi:   {df['time'].iloc[-1] / 60:.2f} dakika")

print("\n=== TEST KÜMESİ SONUÇLARI (VOC2007 Test - 4952 Görsel) ===")
print(f"mAP@0.50:       {test_metrics.box.map50:.4f}")
print(f"mAP@0.50:0.95:  {test_metrics.box.map:.4f}")
print(f"Precision:      {test_metrics.box.mp:.4f}")
print(f"Recall:         {test_metrics.box.mr:.4f}")
print(f"Preprocess:     {speed['preprocess']:.2f} ms/görsel")
print(f"Inference:      {speed['inference']:.2f} ms/görsel")
print(f"Postprocess:    {speed['postprocess']:.2f} ms/görsel")
print(f"Toplam Süre:    {total_ms:.2f} ms/görsel (~{1000 / total_ms:.1f} FPS)")

results_img = os.path.join(RUN_DIR, "results.png")
cm_img = os.path.join(SAVE_DIR, "yolov8n_voc_test", "confusion_matrix_normalized.png")
pr_img = os.path.join(SAVE_DIR, "yolov8n_voc_test", "BoxPR_curve.png")

for path in [results_img, cm_img, pr_img]:
    if os.path.exists(path):
        print(f"\nGrafik: {os.path.basename(path)}")
        display(Image(filename=path, width=800))
```

```python
import os
import gradio as gr
from ultralytics import YOLO

BEST_WEIGHTS = "/content/drive/MyDrive/cv_yolov8_voc/yolov8n_voc_run/weights/best.pt"
deployed_model = YOLO(BEST_WEIGHTS)

def detect_objects(image, conf_threshold, iou_threshold):
    if image is None:
        return None, "Lütfen bir görsel yükleyin."

    results = deployed_model.predict(
        source=image,
        conf=conf_threshold,
        iou=iou_threshold,
        imgsz=640,
        verbose=False,
    )

    res = results[0]
    annotated_rgb = res.plot()[:, :, ::-1]
    inf_ms = res.speed["inference"]
    tot_ms = res.speed["preprocess"] + inf_ms + res.speed["postprocess"]

    boxes = res.boxes
    if len(boxes) == 0:
        report = (
            f"**Inference Süresi:** {inf_ms:.2f} ms (Toplam: {tot_ms:.2f} ms)\n\n"
            "Belirtilen güven eşiğinde nesne tespit edilemedi."
        )
        return annotated_rgb, report

    lines = [
        f"**Inference Süresi:** {inf_ms:.2f} ms (Toplam: {tot_ms:.2f} ms) | **Tespit Sayısı:** {len(boxes)}\n",
        "| # | Sınıf | Güven Skoru | Bounding Box (x1, y1, x2, y2) |",
        "|---|---|---|---|",
    ]
    for idx, box in enumerate(boxes, start=1):
        cls_id = int(box.cls[0])
        cls_name = deployed_model.names[cls_id]
        conf = float(box.conf[0]) * 100
        x1, y1, x2, y2 = [int(v) for v in box.xyxy[0].tolist()]
        lines.append(f"| {idx} | **{cls_name}** | %{conf:.1f} | ({x1}, {y1}, {x2}, {y2}) |")

    return annotated_rgb, "\n".join(lines)

demo = gr.Interface(
    fn=detect_objects,
    inputs=[
        gr.Image(type="pil", label="Görsel Yükle"),
        gr.Slider(minimum=0.1, maximum=0.9, value=0.35, step=0.05, label="Confidence Eşiği"),
        gr.Slider(minimum=0.1, maximum=0.9, value=0.45, step=0.05, label="NMS IoU Eşiği"),
    ],
    outputs=[
        gr.Image(label="Tespit Sonucu"),
        gr.Markdown(label="Tahmin Detayları"),
    ],
    title="YOLOv8n – Pascal VOC Nesne Tespiti",
    description="Pascal VOC (20 sınıf) veri setiyle eğitilmiş YOLOv8n modeli. Fotoğraf yükleyerek anlık tespit sonuçlarını ve tahmin süresini görebilirsiniz.",
)

demo.launch(share=True)
```
